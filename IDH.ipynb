Analisando a distribuição dos valores do IDH

```python
plt.figure(figsize=(10, 6))
sb.histplot(def_IDH['IDHM'], kde=True, bins=30)
plt.title('Distribuição dos Valores de IDHM')
plt.xlabel('IDHM')
plt.ylabel('Frequência')
plt.show()
```

Comparando IDH por Região

```python
media_idh_estado = def_IDH.groupby('Nome da Unidade da Federação')['IDHM'].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 7))
sb.barplot(x=media_idh_estado.index, y=media_idh_estado.values, palette='viridis')
plt.xticks(rotation=90)
plt.xlabel('Estado')
plt.ylabel('IDHM Médio')
plt.title('IDHM Médio por Estado')
plt.tight_layout()
plt.show()
```

```python
def_IDH.describe()
```

```python
# Analisar a relação entre o IDHM e seus componentes
components = ['IDHM Longevidade', 'IDHM Educação', 'IDHM Renda']
plt.figure(figsize=(15, 5))

for i, comp in enumerate(components):
    plt.subplot(1, 3, i + 1)
    sb.scatterplot(x=comp, y='IDHM', data=def_IDH)
    plt.title(f'IDHM vs {comp}')

plt.tight_layout()
plt.show()

# Matriz de correlação

correlation_matrix = def_IDH[['IDHM'] + components].corr()
plt.figure(figsize=(6, 4))
sb.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Matriz de correlação: IDHM e Componentes')
plt.show()
```

```python
# Selecionando colunas relevantes para análise de correlação
colunas_correlacao = [
    'IDHM',
    'Esperança de vida ao nascer',
    'IDHM Educação',
    'IDHM Renda',
    'Subíndice de escolaridade - IDHM Educação',
    'Subíndice de frequência escolar - IDHM Educação',
    'Renda per capita'

]

# Calculando a matriz de correlação
matriz_correlacao = def_IDH[colunas_correlacao].corr()

# Visualizando a matriz de correlação
plt.figure(figsize=(10, 8))
sb.heatmap(matriz_correlacao, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Matriz de Correlação entre IDHM e Indicadores Chave')
plt.show()
```

```python
# Contar o número de estados em cada classe de desenvolvimento
contagem_classes = def_IDHC['classe'].value_counts()

# Visualizar a distribuição das classes de desenvolvimento
plt.figure(figsize=(8, 6))
sb.barplot(x=contagem_classes.index, y=contagem_classes.values, palette='viridis')
plt.title('Distribuição dos Estados por Classe de Desenvolvimento Humano')
plt.xlabel('Classe de Desenvolvimento Humano')
plt.ylabel('Número de Estados')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Encontrar os municípios com os 5 maiores IDHM
top_5_idh = def_IDH.nlargest(5, 'IDHM')

# Encontrar os municípios com os 5 menores IDHM
bottom_5_idh = def_IDH.nsmallest(5, 'IDHM')

print("Municípios com os 5 maiores IDHM:")
display(top_5_idh[['Município', 'Nome da Unidade da Federação', 'IDHM', 'IDHM Educação', 'IDHM Longevidade', 'IDHM Renda']])

print("\nMunicípios com os 5 menores IDHM:")
display(bottom_5_idh[['Município', 'Nome da Unidade da Federação', 'IDHM', 'IDHM Educação', 'IDHM Longevidade', 'IDHM Renda']])
```
